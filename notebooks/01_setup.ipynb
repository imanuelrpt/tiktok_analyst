1. Unduh file dataset dari halaman Kaggle ke komputer kamu (klik **Accept Rules** dulu kalau diminta).

```python
# Sel 3: Upload file dataset ke folder data/
from google.colab import files
import shutil, zipfile

uploaded = files.upload()   # pilih file .csv / .zip dari komputer
print('Berhasil diupload:', list(uploaded.keys()))

for nama in uploaded:
    tujuan = os.path.join(BASE, 'data', nama)
    shutil.move(nama, tujuan)
    if nama.lower().endswith('.zip'):
        with zipfile.ZipFile(tujuan) as z:
            z.extractall(os.path.join(BASE, 'data'))
        print('Zip diekstrak ke data/:', nama)

print('Isi folder data/:', os.listdir(os.path.join(BASE, 'data')))
```

```python
# Sel 6: Fungsi baca CSV yang tahan terhadap masalah encoding
# Dipanggil pada Tahap 2 setelah nama file asli diketahui.
def baca_csv(path, **kwargs):
    for enc in ['utf-8-sig', 'utf-8', 'cp1252', 'latin-1']:
        try:
            df = pd.read_csv(path, encoding=enc, **kwargs)
            print(f'Berhasil: {path} | encoding={enc} | baris={len(df):,} | kolom={df.shape[1]}')
            return df
        except UnicodeDecodeError:
            continue
    raise ValueError('Semua encoding gagal; coba encoding_errors=replace')

# File sangat besar? tambahkan chunksize=100000 lalu gabungkan per potongan.
print('Fungsi baca_csv siap digunakan.')
```
